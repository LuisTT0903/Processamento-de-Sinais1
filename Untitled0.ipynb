```python
[x, Fs] = audioread('handel.wav');
x = x(:,1);  N = length(x);
r = [0.995 0.99 0.90 0.75 0.50];

T    = {fft(x), dct(x)};          % coeficientes das duas transformadas
inv  = {@(Y) real(ifft(Y)), @(Y) idct(Y)};
nome = {'DFT', 'DCT'};

K = zeros(numel(r), 2);  EQM = zeros(numel(r), 2);
for i = 1:numel(r)
    for t = 1:2
        Y  = T{t};
        e  = abs(Y).^2;
        es = sort(e, 'descend');
        k  = find(cumsum(es) >= r(i)*sum(es), 1);   % menor K que atinge r
        mask = e >= es(k)*(1 - 1e-12);              % limiar (preserva pares conjugados)
        xr = inv{t}(Y .* mask);

        K(i,t)   = nnz(mask);
        EQM(i,t) = mean((x - xr).^2);
        audiowrite(sprintf('handel_%s_%g.wav', nome{t}, 100*r(i)), ...
                   xr / max(1, max(abs(xr))), Fs);
    end
    fprintf('r = %5.1f%% | DFT: K = %6d (%.2f%%), EQM = %.3e | DCT: K = %6d (%.2f%%), EQM = %.3e\n', ...
        100*r(i), K(i,1), 100*K(i,1)/N, EQM(i,1), K(i,2), 100*K(i,2)/N, EQM(i,2));
end

figure;
subplot(2,1,1); semilogy(100*r, K, '-o');   grid on;
xlabel('Energia preservada (%)'); ylabel('Coeficientes mantidos'); legend(nome);
subplot(2,1,2); semilogy(100*r, EQM, '-o'); grid on;
xlabel('Energia preservada (%)'); ylabel('EQM'); legend(nome);
```
