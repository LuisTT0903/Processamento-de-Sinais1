```python
#1.
import numpy as np # Importa a biblioteca NumPy, essencial para operações numéricas e array.
import matplotlib.pyplot as plt # Importa Matplotlib para plotar gráficos.
from IPython.display import Audio, display # Importa Audio e display do IPython para reproduzir áudios no notebook.

# --- Parâmetros do sinal ---
duration = 5  # Define a duração do sinal em segundos.
fs = 44100   # Define a frequência de amostragem (fs) em Hertz. 44.1 kHz é uma frequência comum para áudio.
f_values = [500, 5000, 10000] # Lista de frequências (em Hz) para os sinais que serão gerados.

# Vetor de tempo
# Cria um array de tempo 't' que vai de 0 até 'duration' segundos.
# O número de pontos é calculado como fs * duration, garantindo 'fs' amostras por segundo.
# 'endpoint=False' garante que o último ponto não seja incluído, para evitar repetições no caso de sinais periódicos.
t = np.linspace(0, duration, int(fs * duration), endpoint=False)

signals = {} # Dicionário para armazenar os sinais gerados, onde a chave será a frequência.

# --- (a) Gerar gráficos de x(t) no domínio do tempo ---
print("Gerando e plotando sinais:")
for f in f_values: # Itera sobre cada frequência definida em 'f_values'.
    # Gerar o sinal cosseno
    # Calcula o sinal cosseno x(t) = cos(2πft) para a frequência atual 'f'.
    # '2 * np.pi * f * t' calcula o argumento do cosseno em radianos.
    x_t = np.cos(2 * np.pi * f * t)
    signals[f] = x_t # Armazena o sinal gerado no dicionário 'signals' usando a frequência 'f' como chave.

    # Plotar o sinal
    plt.figure(figsize=(12, 4)) # Cria uma nova figura para cada gráfico com um tamanho específico.
    plt.plot(t, x_t) # Plota o sinal x_t em função do tempo t.
    plt.title(f'Sinal x(t) para f = {f} Hz') # Define o título do gráfico, incluindo a frequência atual.
    plt.xlabel('Tempo (s)') # Define o rótulo do eixo X.
    plt.ylabel('Amplitude') # Define o rótulo do eixo Y.
    plt.grid(True) # Adiciona uma grade ao gráfico para melhor legibilidade.
    plt.xlim(0, 0.01) # Define o limite do eixo X para mostrar apenas os primeiros 0.01 segundos do sinal.
                      # Isso é importante porque 5 segundos de sinal com 44.1kHz de amostragem seriam densos demais para visualizar.
    plt.show() # Exibe o gráfico.

# --- (b) Escrever um código para tocar os sinais gerados ---
print("\nReproduzindo os sinais gerados:")
for f, signal in signals.items(): # Itera sobre os sinais armazenados no dicionário 'signals'.
    print(f'Reproduzindo sinal com f = {f} Hz:') # Informa qual sinal está sendo reproduzido.
    # Reproduz o sinal de áudio usando IPython.display.Audio.
    # 'signal' é o array NumPy que contém os dados do áudio.
    # 'rate=fs' define a taxa de amostragem para a reprodução.
    # 'autoplay=False' impede que o áudio comece a tocar automaticamente ao carregar o notebook.
    display(Audio(signal, rate=fs, autoplay=False))
    print('-' * 30) # Imprime uma linha separadora para melhor organização da saída.s
```

o codigo define alguns parametros iniciais para gerar as  frequencias, como tempo de duração e as frequencias exigidas, assim como a frequencia de amostragem e duração para os audios.
Apos isso, ele plota os graficos para visualização e gera os arquivos de audio.

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

# --- Parâmetros do Chirp ---
f0 = 500       # Frequência inicial em Hz
f1 = 10000     # Frequência final em Hz
duration = 5   # Duração em segundos
fs = 44100     # Frequência de amostragem em Hz (44.1 kHz)

# Vetor de tempo
t = np.linspace(0, duration, int(fs * duration), endpoint=False)

chirp_signals = {} # Dicionário para armazenar os chirps gerados

print("Gerando e plotando Chirps:")

# --- (a) Chirp com Varredura de Frequência Linear ---
# Cálculo da fase para varredura linear
phase_linear = 2 * np.pi * (f0 * t + (f1 - f0) / (2 * duration) * t**2)

# Gerar o sinal chirp linear
chirp_linear = np.cos(phase_linear)
chirp_signals['Linear'] = chirp_linear

# Plotar o chirp linear
plt.figure(figsize=(12, 4))
plt.plot(t, chirp_linear)
plt.title(f'Chirp Linear (f0={f0}Hz, f1={f1}Hz)')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.xlim(0, 0.5) # Mostra os primeiros 0.5 segundos para visualizar a variação da frequência
plt.show()

# --- (a) Chirp com Varredura de Frequência Quadrática ---
# Cálculo da fase para varredura quadrática
phase_quadratic = 2 * np.pi * (f0 * t + (f1 - f0) / (3 * duration**2) * t**3)

# Gerar o sinal chirp quadrático
chirp_quadratic = np.cos(phase_quadratic)
chirp_signals['Quadratic'] = chirp_quadratic

# Plotar o chirp quadrático
plt.figure(figsize=(12, 4))
plt.plot(t, chirp_quadratic)
plt.title(f'Chirp Quadrático (f0={f0}Hz, f1={f1}Hz)')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.xlim(0, 0.5) # Mostra os primeiros 0.5 segundos
plt.show()

# --- (a) Chirp com Varredura de Frequência Logarítmica ---
# Parâmetro beta para o chirp logarítmico
beta = np.log(f1 / f0) / duration

# Cálculo da fase para varredura logarítmica
phase_logarithmic = 2 * np.pi * f0 / beta * (np.exp(beta * t) - 1)

# Gerar o sinal chirp logarítmico
chirp_logarithmic = np.cos(phase_logarithmic)
chirp_signals['Logarithmic'] = chirp_logarithmic

# Plotar o chirp logarítmico
plt.figure(figsize=(12, 4))
plt.plot(t, chirp_logarithmic)
plt.title(f'Chirp Logarítmico (f0={f0}Hz, f1={f1}Hz)')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.xlim(0, 0.5) # Mostra os primeiros 0.5 segundos
plt.show()

# --- (b) Reprodução dos Sinais Chirp Gerados ---
print("\nReproduzindo os sinais chirp gerados:")
for chirp_type, signal in chirp_signals.items():
    print(f'Reproduzindo Chirp: {chirp_type}')
    display(Audio(signal, rate=fs, autoplay=False))
    print('-' * 30)
```

**Procedimento para Calcular a Resposta ao Impulso de uma Sala Qualquer (Pergunta 4 do Contexto Original):**

Para calcular a resposta ao impulso de uma sala, o procedimento geral é:

1.  **Excitação do Ambiente:** Gerar um sinal de excitação conhecido e bem definido dentro da sala. Os métodos comuns incluem:
    *   **Impulso Acústico (Explosão):** Um som curto e forte, como um balão estourando, um tiro de partida, ou um estalido. O ideal é que seja o mais próximo possível de um impulso ideal (uma função delta).
    *   **Varredura Senoidal (Sweep):** Um chirp logarítmico (como os gerados anteriormente) que varre uma ampla faixa de frequências. Este é o método mais comum e robusto, pois evita ruídos impulsivos e permite uma melhor relação sinal-ruído.
    *   **Ruído Branco/Rosa:** Emissão de um ruído de banda larga.

2.  **Gravação da Resposta:** Gravar o som resultante (a "resposta") em um ou mais pontos da sala usando um microfone de alta qualidade. É crucial que o microfone esteja em uma posição representativa e que o ambiente esteja o mais silencioso possível para evitar ruídos externos.

3.  **Desconvolução (se necessário):**
    *   Se o sinal de excitação foi um impulso ideal, a gravação já é a resposta ao impulso.
    *   Se foi usada uma varredura senoidal ou ruído, a resposta ao impulso (`h(t)`) é obtida através da **desconvolução** do sinal gravado com o sinal de excitação original. Matematicamente, se $y(t) = x(t) \ast h(t)$, onde $y(t)$ é o sinal gravado e $x(t)$ é o sinal de excitação, então $h(t)$ pode ser encontrado. No domínio da frequência, isso se torna $Y(f) = X(f)H(f)$, então $H(f) = Y(f)/X(f)$, e $h(t)$ é a transformada inversa de Fourier de $H(f)$.

**Limitações e Considerações:**

*   **Ruído:** A presença de ruído ambiente pode contaminar a medição da resposta ao impulso. Usar sinais de excitação com boa relação sinal-ruído e gravar em ambientes controlados minimiza isso.
*   **Acoplamento Fonte-Sala:** A posição da fonte sonora e do microfone na sala afeta a resposta medida. Idealmente, várias posições devem ser testadas para obter uma caracterização mais completa da sala.
*   **Linearidade do Sistema:** Assume-se que a sala é um sistema linear e invariante no tempo. Grandes amplitudes sonoras podem causar efeitos não lineares (por exemplo, vibrações de objetos) que distorcem a medição.
*   **Microfone e Fonte:** A resposta em frequência do microfone e da fonte sonora utilizados deve ser plana (ou conhecida e compensada) para garantir que as características medidas sejam da sala e não dos equipamentos.
*   **Duração do Sinal:** O sinal gravado deve ser longo o suficiente para capturar todo o decaimento da reverberação da sala.
*   **Amostragem:** A frequência de amostragem deve ser alta o suficiente para capturar todas as frequências relevantes presentes na resposta da sala (critério de Nyquist).

## Convolução de Sinais com a Resposta ao Impulso do Banheiro (`h_banheiro.wav`)

Vamos agora calcular a resposta do sinal `h_banheiro.wav` (que representa uma resposta ao impulso `h[n]`) para as entradas `handel.wav` e `sinal_taca.wav`.

### Ajuste de Frequência de Amostragem (Reamostragem)

Para realizar a convolução, todos os sinais devem ter a mesma frequência de amostragem. `handel.wav` tem `fs = 8192 Hz`, enquanto `h_banheiro.wav` e `sinal_taca.wav` têm `fs = 22050 Hz`. Portanto, precisamos reamostrar `handel.wav` para 22050 Hz.

```python
from scipy.signal import resample_poly

# Obter os dados de áudio e as frequências de amostragem
fs_h_banheiro = loaded_signals['h_banheiro.wav']['fs']
h_banheiro_data = loaded_signals['h_banheiro.wav']['data'].astype(float) # Converter para float para convolução

fs_sinal_taca = loaded_signals['sinal_taca.wav']['fs']
sinal_taca_data = loaded_signals['sinal_taca.wav']['data'].astype(float)

# Já temos fs_handel e audio_data (dados de handel.wav) do código anterior
# fs_handel = 8192 Hz
# audio_data = dados de handel.wav (int16)
handel_data_original = audio_data.astype(float)

# Frequência de amostragem alvo para todos os sinais
fs_target = fs_h_banheiro # 22050 Hz

# --- Reamostrar handel.wav ---
# Calcular os fatores de upsampling e downsampling para resample_poly
# Queremos ir de fs_handel (8192) para fs_target (22050)
# up = fs_target / gcd(fs_target, fs_handel)
# down = fs_handel / gcd(fs_target, fs_handel)

# Simplificando a razão: 22050 / 8192 = 11025 / 4096
# up = 11025, down = 4096

# No entanto, resample_poly pode lidar com isso. Vamos usar uma abordagem mais direta com a nova fs
num_samples_handel_resampled = int(len(handel_data_original) * (fs_target / fs_handel))

handel_data_resampled = resample_poly(handel_data_original, fs_target, fs_handel)

print(f"handel.wav original fs: {fs_handel} Hz, amostras: {len(handel_data_original)}")
print(f"handel.wav reamostrado fs: {fs_target} Hz, amostras: {len(handel_data_resampled)}")
print(f"h_banheiro.wav fs: {fs_h_banheiro} Hz, amostras: {len(h_banheiro_data)}")
print(f"sinal_taca.wav fs: {fs_sinal_taca} Hz, amostras: {len(sinal_taca_data)}")
```

### Convolução dos Sinais

Agora que todos os sinais estão na mesma frequência de amostragem, podemos realizar as convoluções.

## Geração e Análise de Chirps (Varredura de Frequência)

Este bloco contém a **explicação e o comentário dos resultados** para os chirps com varreduras linear, quadrática e logarítmica.

### (a) Chirp com Varredura de Frequência Linear

Em um chirp linear, a frequência instantânea muda linearmente com o tempo. A fórmula para a fase instantânea é:

$$\phi(t) = 2\pi \left(f_0 t + \frac{f_1 - f_0}{2T} t^2\right)$$

Onde $T$ é a duração do sinal.

### (a) Chirp com Varredura de Frequência Quadrática

Em um chirp quadrático, a frequência instantânea muda quadraticamente com o tempo. A fórmula para a fase instantânea é:

$$\phi(t) = 2\pi \left(f_0 t + \frac{f_1 - f_0}{3T^2} t^3\right)$$

Onde $T$ é a duração do sinal.

### (a) Chirp com Varredura de Frequência Logarítmica

Em um chirp logarítmico (também conhecido como exponencial), a frequência instantânea muda exponencialmente com o tempo. A fórmula para a fase instantânea é:

$$\phi(t) = 2\pi \frac{f_0 T}{\ln(f_1/f_0)} \left( \left(\frac{f_1}{f_0}\right)^{t/T} - 1 \right)$$

Onde $T$ é a duração do sinal.

### (c) Comentário sobre os Resultados Obtidos

**Análise dos Gráficos (Domínio do Tempo):**

Nos gráficos, observamos a variação da frequência ao longo do tempo. Nos primeiros 0.5 segundos:

*   **Chirp Linear:** A densidade das oscilações (frequência) aumenta de forma constante e gradual. É visível que os ciclos se tornam mais compactos à medida que o tempo avança.
*   **Chirp Quadrático:** A frequência aumenta de forma mais lenta no início e acelera mais bruscamente no final em comparação com o chirp linear. Isso se deve à relação quadrática com o tempo, fazendo com que as frequências mais baixas durem mais e o aumento seja mais acentuado na parte final da varredura.
*   **Chirp Logarítmico:** A frequência aumenta muito rapidamente no início e de forma mais gradual e perceptível em termos proporcionais (logarítmicos) ao longo do tempo. Visualmente, pode parecer que a frequência muda mais rapidamente no começo, mas a taxa de mudança percentual é constante. As variações de frequência em tons mais baixos são mais prolongadas, e as em tons mais altos são mais curtas.

**Análise da Reprodução de Áudio:**

Ao ouvir os sinais, a diferença entre os tipos de varredura se torna mais evidente:

*   **Chirp Linear:** Você ouvirá um som que começa grave (500 Hz) e sobe de tom de maneira suave e constante até atingir um tom agudo (10000 Hz). A transição parece "uniforme" em termos de pitch.
*   **Chirp Quadrático:** O som começará grave e demorará mais para subir em tom no início. A subida de tom será percebida como acelerando, ficando mais rápida à medida que o som se aproxima das frequências mais altas. Parece "empurrar" as frequências mais altas para o final da duração.
*   **Chirp Logarítmico:** Este chirp geralmente soa mais "natural" ou musical para o ouvido humano, pois nossa percepção de pitch é logarítmica. O som também começa grave e sobe agudo, mas a subida de tom é percebida como mais uniforme em termos de "passos" musicais ou oitavas, com a duração das frequências mais baixas sendo relativamente mais longa e as das frequências mais altas mais curtas.

**Conclusão:**

Cada tipo de chirp demonstra uma maneira diferente de variar a frequência ao longo do tempo. Enquanto o chirp linear proporciona uma mudança de frequência diretamente proporcional ao tempo, o quadrático retarda as frequências baixas e acelera as altas, e o logarítmico distribui as frequências de uma forma que se alinha melhor com a percepção auditiva humana, tornando a varredura mais equilibrada em termos de pitch.

```python
#6.
import numpy as np                            # Biblioteca para manipulação numérica e arrays.
import matplotlib.pyplot as plt               # Biblioteca para gerar gráficos.
import urllib.request                         # Para baixar os arquivos de áudio do GitHub.
from scipy.io import wavfile                   # Para ler os arquivos .wav.
from scipy.signal import convolve, resample    # convolve() para convoluir os sinais; resample() caso as taxas sejam diferentes.
from IPython.display import Audio, display     # Para reproduzir os áudios no notebook.

# --- Baixar os arquivos de áudio do repositório no GitHub ---
base_url = 'https://raw.githubusercontent.com/rafaelschaves/gele7317-proc-sin/master/data/'

urllib.request.urlretrieve(base_url + 'h_banheiro.wav', 'h_banheiro.wav')
urllib.request.urlretrieve(base_url + 'handel.wav', 'handel.wav')
urllib.request.urlretrieve(base_url + 'sinal_taca.wav', 'sinal_taca.wav')

print('Arquivos baixados com sucesso!')

# --- Carregar os arquivos de áudio ---
fs_h, h_banheiro = wavfile.read('h_banheiro.wav')     # Lê a resposta ao impulso do banheiro.
fs_handel, handel = wavfile.read('handel.wav')        # Lê o sinal de áudio (música).
fs_taca, taca = wavfile.read('sinal_taca.wav')        # Lê o sinal da taça.

# --- Conferir as frequências de amostragem ---
print(f'fs h_banheiro: {fs_h} Hz')
print(f'fs handel: {fs_handel} Hz')
print(f'fs taca: {fs_taca} Hz')

# Caso as frequências sejam diferentes, é necessário reamostrar (resample) os sinais
# para que fiquem na mesma taxa antes da convolução, conforme a dica do enunciado.
def igualar_fs(sinal, fs_original, fs_alvo):
    if fs_original != fs_alvo:
        n_amostras = int(len(sinal) * fs_alvo / fs_original)   # Novo número de amostras proporcional à nova fs.
        sinal = resample(sinal, n_amostras)                     # Reamostra o sinal para a nova taxa.
    return sinal

# Define a fs do h_banheiro como referência (poderia ser qualquer uma, desde que todas fiquem iguais).
fs_alvo = fs_h
handel = igualar_fs(handel, fs_handel, fs_alvo)
taca = igualar_fs(taca, fs_taca, fs_alvo)

# Caso os sinais sejam estéreo (2 colunas), converte para mono somando/tirando a média dos canais.
if h_banheiro.ndim > 1:
    h_banheiro = h_banheiro.mean(axis=1)
if handel.ndim > 1:
    handel = handel.mean(axis=1)
if taca.ndim > 1:
    taca = taca.mean(axis=1)

# Normaliza os sinais (opcional, mas evita estouro de amplitude na convolução).
h_banheiro = h_banheiro / np.max(np.abs(h_banheiro))
handel = handel / np.max(np.abs(handel))
taca = taca / np.max(np.abs(taca))

# --- (a) Calcular a convolução (resposta do sistema h_banheiro para cada entrada) ---
saida_handel = convolve(handel, h_banheiro)    # y[n] = x[n] * h[n], sinal handel passando pelo banheiro.
saida_taca = convolve(taca, h_banheiro)        # y[n] = x[n] * h[n], sinal da taça passando pelo banheiro.

# Normaliza as saídas para evitar clipping na reprodução.
saida_handel = saida_handel / np.max(np.abs(saida_handel))
saida_taca = saida_taca / np.max(np.abs(saida_taca))

# --- Avaliar as respostas no tempo ---
t_handel = np.arange(len(saida_handel)) / fs_alvo
t_taca = np.arange(len(saida_taca)) / fs_alvo

plt.figure(figsize=(12, 4))
plt.plot(t_handel, saida_handel)
plt.title('Convolução: handel.wav * h_banheiro[n]')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()

plt.figure(figsize=(12, 4))
plt.plot(t_taca, saida_taca)
plt.title('Convolução: sinal_taca.wav * h_banheiro[n]')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()

# --- (b) Ouvir as respostas ---
print('Handel convoluído com h_banheiro:')
display(Audio(saida_handel, rate=fs_alvo, autoplay=False))

print('Taça convoluída com h_banheiro:')
display(Audio(saida_taca, rate=fs_alvo, autoplay=False))
```
