A transformada de Fourier representa um sinal como uma soma de senoides de duração infinita. Ela mostra com precisão quais frequências estão presentes, mas não informa em que instante cada uma ocorre. Isso não é um problema para sinais estacionários, mas a maioria dos sinais reais não é estacionária: a fala, um eletrocardiograma, um transitório em uma linha de transmissão e uma onda sísmica mudam de comportamento ao longo do tempo. Uma primeira solução é a transformada de Fourier de tempo curto (STFT), que aplica a transformada de Fourier a trechos do sinal recortados por uma janela que desliza no tempo. A limitação da STFT é que a largura da janela é fixa. Pelo princípio da incerteza, uma janela curta dá boa localização no tempo e má resolução em frequência, e uma janela longa faz o contrário. Com a STFT, é preciso escolher um único compromisso para todas as frequências.

As wavelets resolvem esse problema usando, no lugar das senoides, pequenas ondas de duração limitada e média nula. Todas elas são geradas a partir de uma única função, chamada wavelet mãe, por meio de duas operações: a dilatação (mudança de escala) e a translação (deslocamento no tempo). A transformada wavelet mede a semelhança entre o sinal e cada versão dilatada e deslocada da wavelet mãe. Em escalas pequenas, a wavelet é comprimida, oscila rápido e dura pouco, o que permite analisar componentes de alta frequência com ótima localização no tempo. Em escalas grandes, a wavelet é esticada, oscila devagar e dura mais, o que permite analisar componentes de baixa frequência com boa resolução em frequência. Assim, a resolução se adapta automaticamente: detalhes rápidos são localizados com precisão no tempo, e tendências lentas são medidas com precisão em frequência. Essa é a chamada análise multirresolução.

Existem duas formas principais da transformada. A transformada wavelet contínua (CWT) calcula os coeficientes para todas as escalas e posições. O resultado é um mapa tempo-escala, chamado escalograma, muito útil para visualizar e interpretar sinais, mas bastante redundante. A transformada wavelet discreta (DWT) usa apenas escalas e posições em potências de 2, o que elimina a redundância e permite reconstruir o sinal exatamente. Em 1989, Stéphane Mallat mostrou que a DWT pode ser calculada por um banco de filtros. Em cada nível, o sinal passa por um filtro passa-baixas e um filtro passa-altas e depois é subamostrado por 2. A saída passa-baixas é a aproximação, que contém a tendência do sinal, e a saída passa-altas é o detalhe, que contém as variações rápidas. A aproximação é decomposta novamente no nível seguinte, formando uma árvore. Esse algoritmo tem custo computacional proporcional ao número de amostras, menor até que o da FFT.

Historicamente, a primeira wavelet foi proposta por Alfréd Haar em 1910, embora o termo só tenha surgido décadas depois. A teoria moderna nasceu no início dos anos 1980, com o geofísico Jean Morlet e o físico Alex Grossmann, que buscavam analisar sinais sísmicos na prospecção de petróleo. Em seguida, Yves Meyer formalizou a teoria no contexto da análise funcional, Mallat estabeleceu a ligação com os bancos de filtros, e Ingrid Daubechies construiu, em 1988, as primeiras famílias de wavelets ortogonais de suporte compacto. Hoje existem diversas famílias, cada uma com propriedades diferentes de suavidade, simetria e duração: Haar, Daubechies, Symlets, Coiflets, biortogonais, Morlet e chapéu mexicano, entre outras. A escolha da wavelet depende do sinal analisado: em geral, procura-se uma wavelet cuja forma se pareça com os padrões que se quer detectar.

As aplicações são muito variadas. Em compressão de dados, a maior parte da energia de sinais e imagens naturais se concentra em poucos coeficientes wavelet, e os demais podem ser descartados ou quantizados com pouca perda de qualidade. Esse é o princípio do padrão de compressão de imagens JPEG 2000 e do sistema usado pelo FBI para armazenar impressões digitais. Na remoção de ruído, a técnica de limiarização proposta por Donoho zera os coeficientes de pequena amplitude, que correspondem principalmente ao ruído, e preserva os grandes, que correspondem ao sinal. Diferente de um filtro passa-baixas, essa técnica remove o ruído sem suavizar bordas e transitórios importantes.

Na engenharia elétrica, as wavelets são usadas na análise de qualidade de energia e na proteção de sistemas de potência, para detectar e localizar faltas, afundamentos de tensão, harmônicos e transitórios de chaveamento, graças à boa localização temporal dos coeficientes de detalhe. Também são usadas no monitoramento de vibrações e no diagnóstico de falhas em máquinas rotativas, rolamentos e engrenagens. Na área biomédica, são aplicadas na análise de ECG, para detectar o complexo QRS, e de EEG, para identificar padrões de crises epilépticas. Na geofísica, continuam sendo usadas na análise de dados sísmicos, área que deu origem à teoria. Em telecomunicações, há esquemas de modulação multiportadora baseados em wavelets, como o adotado no padrão IEEE 1901 para comunicação pela rede elétrica. Outras aplicações incluem o reconhecimento de voz, a detecção de bordas e texturas em imagens, a análise de séries financeiras e a solução numérica de equações diferenciais.

Em resumo, as transformadas wavelet são uma ferramenta de análise tempo-frequência que complementa a transformada de Fourier. Enquanto Fourier é ideal para sinais estacionários, as wavelets são mais adequadas para sinais com transitórios, descontinuidades e características que mudam ao longo do tempo. Sua capacidade de representar sinais em várias resoluções, aliada a algoritmos rápidos de cálculo, explica sua ampla adoção em compressão, filtragem, detecção de eventos e diagnóstico.
