```python
import os
import wave
import urllib.request
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display

url = "https://raw.githubusercontent.com/LuisTT0903/Processamento-de-Sinais1/main/Aula_07/Audios_Usados/handel.wav"
if not os.path.exists("handel.wav"):
    urllib.request.urlretrieve(url, "handel.wav")

with wave.open("handel.wav", "rb") as w:
    fs = w.getframerate()
    x = np.frombuffer(w.readframes(w.getnframes()), dtype=np.int16).astype(float) / 32768.0
N = len(x)
t = np.arange(N) / fs
i1 = 0.05 * np.cos(200 * np.pi * t)
i2 = 0.075 * np.sin(4000 * np.pi * t)
interf = i1 + i2
variancias = [1e-2, 1e-1, 1.0]
rng = np.random.default_rng(0)
ruidos = {s2: np.sqrt(s2) * rng.standard_normal(N) for s2 in variancias}
ys = {s2: x + interf + ruidos[s2] for s2 in variancias}


# ---------------------------------------------------------------------------
# Funções auxiliares
# ---------------------------------------------------------------------------
def pot(v):
    return np.mean(v ** 2)


def hann(n):
    return 0.5 - 0.5 * np.cos(2 * np.pi * np.arange(n) / n)


def psd(v, nperseg=1024):
    # Densidade espectral pelo método de Welch: janelas de Hann com 50% de
    # sobreposição, média dos periodogramas, escala unilateral em V^2/Hz
    w = hann(nperseg)
    passo = nperseg // 2
    P = np.zeros(nperseg // 2 + 1)
    K = 0
    for i in range(0, len(v) - nperseg + 1, passo):
        s = v[i:i + nperseg]
        P += np.abs(np.fft.rfft((s - s.mean()) * w)) ** 2
        K += 1
    P /= K * fs * np.sum(w ** 2)
    P[1:-1] *= 2
    return np.arange(len(P)) * fs / nperseg, 10 * np.log10(P + 1e-20)


def espectrograma(ax, v, titulo, nfft=512, sobrep=384):
    w = hann(nfft)
    passo = nfft - sobrep
    quadros = [np.abs(np.fft.rfft(v[i:i + nfft] * w)) ** 2 for i in range(0, len(v) - nfft + 1, passo)]
    S = np.array(quadros).T / (fs * np.sum(w ** 2))
    S[1:-1] *= 2
    tq = (np.arange(S.shape[1]) * passo + nfft / 2) / fs
    fq = np.arange(S.shape[0]) * fs / nfft
    ax.pcolormesh(tq, fq, 10 * np.log10(S + 1e-20), cmap="magma", vmin=-120, vmax=-20, shading="auto")
    ax.set_title(titulo)
    ax.set_xlabel("t [s]")


def resp_freq(b, a, f):
    z = np.exp(2j * np.pi * np.asarray(f) / fs)
    return np.polyval(b, z) / np.polyval(a, z)


def resp_cascata(sos_, f):
    H = np.ones(len(f), complex)
    for s in sos_:
        H *= resp_freq(s[:3], s[3:], f)
    return H


def raizes2(c):
    c0, c1, c2 = c
    d = np.sqrt(complex(c1 ** 2 - 4 * c0 * c2))
    return np.array([(-c1 + d) / (2 * c0), (-c1 - d) / (2 * c0)])


def filtra_bloco(s, v):
    # y[n] = b0 x[n] + b1 x[n-1] + b2 x[n-2] - a1 y[n-1] - a2 y[n-2]
    b0, b1, b2 = s[0] / s[3], s[1] / s[3], s[2] / s[3]
    a1, a2 = s[4] / s[3], s[5] / s[3]
    y = [0.0] * len(v)
    x1 = x2 = y1 = y2 = 0.0
    for n, xn in enumerate(np.asarray(v, float).tolist()):
        yn = b0 * xn + b1 * x1 + b2 * x2 - a1 * y1 - a2 * y2
        x2, x1, y2, y1 = x1, xn, y1, yn
        y[n] = yn
    return np.array(y)


def filtra_cascata(sos_, v):
    for s in sos_:
        v = filtra_bloco(s, v)
    return v


def filtra_fase_zero(sos_, v):
    # Filtra para frente e depois de trás para frente: fase nula, módulo ao quadrado
    return filtra_cascata(sos_, filtra_cascata(sos_, v)[::-1])[::-1]


def filtra_direta(b, a, v):
    # Equação a diferenças do filtro completo (forma direta), de ordem qualquer
    b = np.asarray(b, float) / a[0]
    a = np.asarray(a, float) / a[0]
    M = len(b)
    xb, yb = [0.0] * M, [0.0] * M
    y = [0.0] * len(v)
    bl, al = b.tolist(), a.tolist()
    for n, xn in enumerate(np.asarray(v, float).tolist()):
        xb = [xn] + xb[:-1]
        yn = sum(bk * xk for bk, xk in zip(bl, xb)) - sum(ak * yk for ak, yk in zip(al[1:], yb[:-1]))
        yb = [yn] + yb[:-1]
        y[n] = yn
    return np.array(y)


def sos_para_tf(sos_):
    b_, a_ = np.array([1.0]), np.array([1.0])
    for s in sos_:
        b_, a_ = np.convolve(b_, s[:3]), np.convolve(a_, s[3:])
    return b_, a_


def bloco_pb(fp, r):
    # Bloco passa-baixas básico: H(z) = G (z + 1)^2 / (z^2 - 2 r cos(theta) z + r^2), H(1) = 1
    th = 2 * np.pi * fp / fs
    a = np.array([1.0, -2 * r * np.cos(th), r ** 2])
    b = np.array([1.0, 2.0, 1.0])
    return np.r_[np.polyval(a, 1.0) / np.polyval(b, 1.0) * b, a]


def notch(f0, r):
    # Bloco notch básico: zeros em e^(±j theta), polos em r e^(±j theta)
    th = 2 * np.pi * f0 / fs
    b = np.array([1.0, -2 * np.cos(th), 1.0])
    a = np.array([1.0, -2 * r * np.cos(th), r ** 2])
    z_ref = -1.0 if f0 < fs / 4 else 1.0
    g = np.polyval(a, z_ref) / np.polyval(b, z_ref)
    return np.r_[g * b, a]


def passa_baixas(fc, ordem):
    # Cascata de blocos passa-baixas básicos com polos posicionados como num
    # Butterworth: polos analógicos s_k = Wc e^(j pi (2k + N + 1) / 2N) levados ao
    # plano z pela transformação bilinear z = (2 fs + s) / (2 fs - s), com pré-distorção
    # Wc = 2 fs tan(pi fc / fs). Cada par conjugado vira um bloco com
    # r = |z_k| e theta = ang(z_k); os zeros do bloco ficam em z = -1.
    Wc = 2 * fs * np.tan(np.pi * fc / fs)
    blocos = []
    for k in range(ordem // 2):
        s_k = Wc * np.exp(1j * np.pi * (2 * k + ordem + 1) / (2 * ordem))
        z_k = (2 * fs + s_k) / (2 * fs - s_k)
        blocos.append((abs(z_k), abs(np.angle(z_k)) * fs / (2 * np.pi)))
    blocos.sort()
    return np.array([bloco_pb(fp, r) for r, fp in blocos]), blocos


# ---------------------------------------------------------------------------
print(f"fs = {fs} Hz, N = {N} amostras, duração = {N / fs:.2f} s")
print(f"Potência de x: {pot(x):.5f}   interferência 100 Hz: {pot(i1):.5f}   interferência 2 kHz: {pot(i2):.5f}")

print("\n(a) Sinal contaminado")
for s2 in variancias:
    print(f"sigma^2 = {s2:<5}  potência do ruído = {pot(ruidos[s2]):.4f}   SNR de entrada = {10 * np.log10(pot(x) / pot(ys[s2] - x)):6.2f} dB")

n0, n1 = int(2.0 * fs), int(2.05 * fs)
fig, ax = plt.subplots(len(variancias), 2, figsize=(15, 10))
for k, s2 in enumerate(variancias):
    ax[k, 0].plot(t[n0:n1] * 1000, ys[s2][n0:n1], lw=0.8, label="y(t)")
    ax[k, 0].plot(t[n0:n1] * 1000, x[n0:n1], lw=1.5, label="x(t)")
    ax[k, 0].set_title(f"Domínio do tempo (trecho de 50 ms) - sigma^2 = {s2}")
    ax[k, 0].set_xlabel("t [ms]")
    ax[k, 0].legend(loc="upper right")
    fx, Px = psd(x)
    fy, Py = psd(ys[s2])
    ax[k, 1].plot(fy, Py, lw=0.9, label="y(t)")
    ax[k, 1].plot(fx, Px, lw=0.9, label="x(t)")
    ax[k, 1].set_title(f"Densidade espectral de potência - sigma^2 = {s2}")
    ax[k, 1].set_xlabel("f [Hz]")
    ax[k, 1].set_ylabel("dB/Hz")
    ax[k, 1].legend(loc="upper right")
for a_ in ax.flat:
    a_.grid(alpha=0.3)
fig.tight_layout()
plt.savefig("Q5_01.png", dpi=110)
plt.show()

fig, ax = plt.subplots(1, 4, figsize=(18, 4), sharey=True)
for a_, (nome, v) in zip(ax, [("x(t) original", x)] + [(f"y(t), sigma^2 = {s2}", ys[s2]) for s2 in variancias]):
    espectrograma(a_, v, nome)
ax[0].set_ylabel("f [Hz]")
fig.tight_layout()
plt.savefig("Q5_02.png", dpi=110)
plt.show()

print("Áudio original:")
display(Audio(x, rate=fs, autoplay=False))
for s2 in variancias:
    print(f"Áudio contaminado, sigma^2 = {s2}:")
    display(Audio(ys[s2], rate=fs, autoplay=False))

print("\n(b) Análise espectral")
X = np.abs(np.fft.rfft(x)) ** 2
fX = np.arange(len(X)) * fs / N
acum = np.cumsum(X) / X.sum()
for p in [0.90, 0.95, 0.99]:
    print(f"{100 * p:.0f}% da energia de x(t) abaixo de {fX[np.searchsorted(acum, p)]:.0f} Hz")
for s2 in variancias:
    print(f"sigma^2 = {s2}: densidade do ruído = {10 * np.log10(2 * s2 / fs):.1f} dB/Hz (unilateral); "
          f"ruído acima de 1800 Hz = {100 * (fs / 2 - 1800) / (fs / 2):.0f}% da potência total do ruído")

fig, ax = plt.subplots(figsize=(14, 5))
fx, Px = psd(x)
ax.plot(fx, Px, "k", lw=1.5, label="x(t)")
for s2 in variancias:
    fy, Py = psd(ys[s2])
    ax.plot(fy, Py, lw=0.9, label=f"y(t), sigma^2 = {s2}")
    ax.axhline(10 * np.log10(2 * s2 / fs), ls=":", color="gray")
for fi, nome in [(100, "100 Hz"), (2000, "2000 Hz")]:
    ax.axvline(fi, color="r", ls="--", lw=0.8)
    ax.text(fi + 30, -15, nome, color="r")
ax.axvspan(0, 1800, color="green", alpha=0.06, label="faixa útil escolhida (0-1800 Hz)")
ax.set_xlabel("f [Hz]")
ax.set_ylabel("dB/Hz")
ax.set_title("Espectro do sinal contaminado (linhas pontilhadas: nível teórico do ruído branco)")
ax.grid(alpha=0.3)
ax.legend()
plt.savefig("Q5_03.png", dpi=110)
plt.show()

r_notch = 0.99
fc_pb = 1800
ordem_pb = 6
sos_n1 = notch(100, r_notch)[None, :]
sos_n2 = notch(2000, r_notch)[None, :]
sos_pb, par_pb = passa_baixas(fc_pb, ordem_pb)
sos = np.vstack([sos_n1, sos_n2, sos_pb])
nomes = ["Notch 100 Hz", "Notch 2000 Hz"] + [f"Passa-baixas bloco {k + 1}" for k in range(len(sos_pb))]

print("\n(c) Sistema de filtragem: cascata de", len(sos), "blocos de 2ª ordem")
print(f"Passa-baixas: {ordem_pb // 2} blocos básicos (z + 1)^2 / (z^2 - 2r cos(theta) z + r^2), fc = {fc_pb} Hz")
for k, (r, fp) in enumerate(par_pb, 1):
    print(f"    bloco {k}: polos em r = {r:.4f}, theta = 2 pi ({fp:.1f} Hz) / fs")
for nome, s in zip(nomes, sos):
    b, a = s[:3], s[3:]
    z, p = raizes2(b), raizes2(a)
    print(f"{nome}: H(z) = ({b[0]:.6f} z^2 {b[1]:+.6f} z {b[2]:+.6f}) / (z^2 {a[1]:+.6f} z {a[2]:+.6f})")
    print(f"    polos: r = {np.abs(p[0]):.4f}, ângulo = {abs(np.angle(p[0])) * fs / (2 * np.pi):.1f} Hz   "
          f"zeros: |z| = {np.abs(z[0]):.4f}, ângulo = {abs(np.angle(z[0])) * fs / (2 * np.pi):.1f} Hz")

fw = np.linspace(0, fs / 2, 8192, endpoint=False)
Htot = resp_cascata(sos, fw)
blocos_resp = [resp_cascata(s[None, :], fw) for s in sos]
fig, ax = plt.subplots(2, 2, figsize=(15, 10))
for nome, H in zip(nomes, blocos_resp):
    ax[0, 0].plot(fw, 20 * np.log10(np.abs(H) + 1e-12), lw=1, label=nome)
ax[0, 0].plot(fw, 20 * np.log10(np.abs(Htot) + 1e-12), "k", lw=2, label="Sistema completo")
ax[0, 0].set_ylim(-80, 10)
ax[0, 0].set_title("Módulo da resposta em frequência")
ax[0, 0].set_xlabel("f [Hz]")
ax[0, 0].set_ylabel("dB")
ax[0, 0].legend(fontsize=8)
ax[0, 1].plot(fw, np.unwrap(np.angle(Htot)), "k")
ax[0, 1].set_title("Fase do sistema completo")
ax[0, 1].set_xlabel("f [Hz]")
ax[0, 1].set_ylabel("rad")
th = np.linspace(0, 2 * np.pi, 400)
ax[1, 0].plot(np.cos(th), np.sin(th), "k--", lw=0.8)
cores = plt.cm.tab10(np.arange(len(sos)))
for nome, s, c in zip(nomes, sos, cores):
    z, p = raizes2(s[:3]), raizes2(s[3:])
    ax[1, 0].plot(p.real, p.imag, "x", color=c, ms=10, mew=2, label=nome)
    ax[1, 0].plot(z.real, z.imag, "o", mfc="none", mec=c, ms=10, mew=1.5)
ax[1, 0].set_aspect("equal")
ax[1, 0].set_title("Polos (x) e zeros (o) de todos os blocos")
ax[1, 0].legend(fontsize=7, loc="upper left", bbox_to_anchor=(1.02, 1))
ax[1, 1].plot(fw, 20 * np.log10(np.abs(Htot) + 1e-12), "k")
ax[1, 1].set_xlim(0, 300)
ax[1, 1].set_ylim(-60, 5)
ax[1, 1].set_title("Detalhe do notch em 100 Hz")
ax[1, 1].set_xlabel("f [Hz]")
for a_ in ax.flat:
    a_.grid(alpha=0.3)
fig.tight_layout()
plt.savefig("Q5_04.png", dpi=110)
plt.show()

fig, ax = plt.subplots(1, 2, figsize=(15, 4.5))
fcs = np.arange(600, 3801, 200)
sos_notches = np.vstack([sos_n1, sos_n2])
for s2 in variancias:
    y_n = filtra_fase_zero(sos_notches, ys[s2])
    snr = []
    for fc in fcs:
        snr.append(10 * np.log10(pot(x) / pot(filtra_fase_zero(passa_baixas(fc, ordem_pb)[0], y_n) - x)))
    ax[0].plot(fcs, snr, "o-", label=f"sigma^2 = {s2}")
    print(f"sigma^2 = {s2}: frequência de corte com maior SNR = {fcs[np.argmax(snr)]} Hz")
ax[0].axvline(fc_pb, color="k", ls=":")
ax[0].set_title("SNR de saída x frequência de corte do passa-baixas (fase zero)")
ax[0].set_xlabel("fc [Hz]")
ax[0].set_ylabel("SNR [dB]")
ax[0].legend()
rs = [0.9, 0.95, 0.98, 0.99, 0.995, 0.999]
dist, resid = [], []
for r in rs:
    s_t = np.vstack([notch(100, r), notch(2000, r)])
    dist.append(10 * np.log10(pot(x) / pot(filtra_cascata(s_t, x) - x)))
    resid.append(10 * np.log10(pot(filtra_cascata(s_t, interf)) / pot(interf)))
ax[1].plot(rs, dist, "o-", label="SNR de distorção do sinal pelos notches [dB]")
ax[1].plot(rs, resid, "s-", label="interferência residual (transitório) [dB]")
ax[1].axvline(r_notch, color="k", ls=":")
ax[1].set_title("Efeito de r nos notches")
ax[1].set_xlabel("r")
ax[1].legend()
for a_ in ax:
    a_.grid(alpha=0.3)
fig.tight_layout()
plt.savefig("Q5_05.png", dpi=110)
plt.show()
for r, d, e in zip(rs, dist, resid):
    print(f"r = {r}: distorção do sinal pelos notches -> SNR {d:6.2f} dB   interferência residual {e:7.2f} dB   "
          f"largura do notch ~ {(1 - r) * fs / np.pi:.1f} Hz")


def melhor_atraso(ref, v, max_lag=60):
    erros = [pot(v[L:] - ref[:N - L]) for L in range(max_lag)]
    L = int(np.argmin(erros))
    return L, erros[L]


x_filt = filtra_cascata(sos, x)
L0, _ = melhor_atraso(x, x_filt)
print(f"\nAtraso médio do sistema (alinhamento por mínimo erro): {L0} amostras = {1000 * L0 / fs:.2f} ms")

xr = {s2: filtra_cascata(sos, ys[s2]) for s2 in variancias}

print("\n(d) Comparação temporal e espectral")
fig, ax = plt.subplots(len(variancias), 2, figsize=(15, 10))
for k, s2 in enumerate(variancias):
    xa = np.r_[xr[s2][L0:], np.zeros(L0)]
    ax[k, 0].plot(t[n0:n1] * 1000, ys[s2][n0:n1], lw=0.6, alpha=0.6, label="y(t)")
    ax[k, 0].plot(t[n0:n1] * 1000, xa[n0:n1], lw=1.3, label="x_rec(t) (alinhado)")
    ax[k, 0].plot(t[n0:n1] * 1000, x[n0:n1], "k", lw=1.3, label="x(t)")
    ax[k, 0].set_title(f"Tempo - sigma^2 = {s2}")
    ax[k, 0].set_xlabel("t [ms]")
    ax[k, 0].legend(loc="upper right", fontsize=8)
    for v, nome in [(ys[s2], "y(t)"), (xr[s2], "x_rec(t)"), (x, "x(t)")]:
        fv, Pv = psd(v)
        ax[k, 1].plot(fv, Pv, lw=0.9 if nome != "x(t)" else 1.3, label=nome, color="k" if nome == "x(t)" else None)
    ax[k, 1].set_title(f"Densidade espectral - sigma^2 = {s2}")
    ax[k, 1].set_xlabel("f [Hz]")
    ax[k, 1].legend(loc="upper right", fontsize=8)
for a_ in ax.flat:
    a_.grid(alpha=0.3)
fig.tight_layout()
plt.savefig("Q5_06.png", dpi=110)
plt.show()

fig, ax = plt.subplots(1, 4, figsize=(18, 4), sharey=True)
for a_, (nome, v) in zip(ax, [("x(t) original", x)] + [(f"x_rec(t), sigma^2 = {s2}", xr[s2]) for s2 in variancias]):
    espectrograma(a_, v, nome)
ax[0].set_ylabel("f [Hz]")
fig.tight_layout()
plt.savefig("Q5_07.png", dpi=110)
plt.show()

print("\n(e) Avaliação quantitativa")
k0 = int(0.1 * fs)
interf_filt = filtra_cascata(sos, interf)


def metricas(s2):
    L, mse = melhor_atraso(x, xr[s2])
    snr_out = 10 * np.log10(pot(x) / mse)
    v0 = filtra_fase_zero(sos, ys[s2])
    snr_0 = 10 * np.log10(pot(x) / pot(v0 - x))
    p_ruido = pot(filtra_cascata(sos, ruidos[s2]))
    p_int = pot(interf_filt[k0:])
    p_dist = pot(x_filt[L:] - x[:N - L])
    return snr_out, mse, snr_0, p_ruido, p_int, p_dist


print(f"{'sigma^2':>8} {'SNR ent.':>9} {'MSE ent.':>9} {'SNR saída':>10} {'MSE saída':>10} {'ganho':>7} "
      f"{'SNR fase 0':>11} {'P ruído ent.':>13} {'P ruído res.':>13} {'P interf. res.':>15} {'P distorção':>12}")
snr_saida = {}
for s2 in variancias:
    snr_in = 10 * np.log10(pot(x) / pot(ys[s2] - x))
    snr_out, mse, snr_0, p_r, p_i, p_d = metricas(s2)
    snr_saida[s2] = snr_out
    print(f"{s2:>8} {snr_in:>9.2f} {pot(ys[s2] - x):>9.4f} {snr_out:>10.2f} {mse:>10.4f} {snr_out - snr_in:>7.2f} "
          f"{snr_0:>11.2f} {pot(ruidos[s2]):>13.4f} {p_r:>13.4f} {p_i:>15.2e} {p_d:>12.5f}")
print(f"Potência das interferências na entrada: {pot(interf):.2e}")

print("\n(f) Áudios recuperados")
for s2 in variancias:
    print(f"Recuperado, sigma^2 = {s2}:")
    display(Audio(xr[s2], rate=fs, autoplay=False))

print("\n(g) Quantização dos coeficientes")


def quantiza(c, b, monico=False):
    # Ponto fixo com b bits; escala do vetor = potência de 2 que cobre o maior coeficiente
    c = np.asarray(c, float)
    q = c.copy()
    idx = slice(1, None) if monico else slice(None)
    v = c[idx]
    m = np.max(np.abs(v))
    if m == 0:
        return q
    inteiros = int(np.floor(np.log2(m))) + 1
    passo = 2.0 ** (inteiros - (b - 1))
    q[idx] = np.clip(np.round(v / passo) * passo, -2.0 ** inteiros, 2.0 ** inteiros - passo)
    return q


def quantiza_cascata(sos_, nb):
    return np.array([np.r_[quantiza(s[:3], nb), quantiza(s[3:], nb, monico=True)] for s in sos_])


b_dir, a_dir = sos_para_tf(sos)
print(f"Forma direta do sistema completo: ordem {len(a_dir) - 1}")
print("a =", np.round(a_dir, 5))
fq = np.linspace(1, fs / 2 - 1, 4000)
H0 = np.abs(resp_freq(b_dir, a_dir, fq))
banda = ((fq > 300) & (fq < 1700)) & (np.abs(fq - 2000) > 200)
bits = [2, 4, 8, 16]
s2_ref = 1e-2
print(f"SNR de saída do sistema original (sigma^2 = {s2_ref}): {snr_saida[s2_ref]:.2f} dB")
print("Original: |H| em 100 Hz e 2 kHz = -inf dB (zeros sobre o círculo unitário)")
print(f"{'forma':>8} {'b':>3} {'max|p|':>9} {'estável':>8} {'erro máx 300-1700 Hz [dB]':>26} {'|H| 100 Hz':>11} {'|H| 2 kHz':>10} {'SNR saída':>10}")
res_q = {"cascata": [], "direta": []}
for forma in ["cascata", "direta"]:
    for nb in bits:
        if forma == "cascata":
            sq = quantiza_cascata(sos, nb)
            pq = np.concatenate([raizes2(s[3:]) for s in sq])
            bq, aq = sos_para_tf(sq)
            Hq = np.abs(resp_cascata(sq, fq))
        else:
            bq, aq = quantiza(b_dir, nb), quantiza(a_dir, nb, monico=True)
            pq = np.roots(aq)
            Hq = np.abs(resp_freq(bq, aq, fq))
        zq = np.roots(bq) if np.count_nonzero(bq) > 1 else np.array([])
        mp = np.max(np.abs(pq))
        estado = "sim" if mp < 1 - 1e-6 else ("marg." if mp <= 1 + 1e-6 else "NÃO")
        erro = np.max(np.abs(20 * np.log10(Hq[banda] + 1e-12) - 20 * np.log10(H0[banda])))
        H_t = np.abs(resp_freq(bq, aq, [100, 2000]))
        h100, h2k = 20 * np.log10(H_t + 1e-12)
        if estado == "sim":
            saida = filtra_cascata(sq, ys[s2_ref]) if forma == "cascata" else filtra_direta(bq, aq, ys[s2_ref])
            L, mse = melhor_atraso(x, saida)
            snr_q = 10 * np.log10(pot(x) / mse) if np.isfinite(mse) else np.nan
        else:
            saida, snr_q = None, np.nan
        res_q[forma].append((nb, pq, zq, Hq, estado, saida))
        print(f"{forma:>8} {nb:>3} {mp:>9.5f} {estado:>8} {erro:>26.2f} {h100:>11.1f} {h2k:>10.1f} "
              f"{(f'{snr_q:.2f}' if np.isfinite(snr_q) else ('marginal' if estado == 'marg.' else 'instável')):>10}")

fig, ax = plt.subplots(1, 2, figsize=(15, 5), sharey=True)
for k, forma in enumerate(["cascata", "direta"]):
    ax[k].plot(fq, 20 * np.log10(H0), "k", lw=3, alpha=0.4, label="original")
    for (nb, pq, zq, Hq, estado, saida), c in zip(res_q[forma], plt.cm.plasma(np.linspace(0, 0.85, len(bits)))):
        ax[k].plot(fq, 20 * np.log10(Hq + 1e-12), color=c, lw=1.2, ls="-" if estado == "sim" else ":",
                   label=f"b = {nb}" + {"sim": "", "marg.": " (marg. estável)", "NÃO": " (instável)"}[estado])
    ax[k].set_ylim(-80, 30)
    ax[k].set_title(f"Coeficientes quantizados - forma {forma}")
    ax[k].set_xlabel("f [Hz]")
    ax[k].grid(alpha=0.3)
    ax[k].legend(fontsize=8)
ax[0].set_ylabel("|H| [dB]")
fig.tight_layout()
plt.savefig("Q5_08.png", dpi=110)
plt.show()

p0 = np.concatenate([raizes2(s[3:]) for s in sos])
z0 = np.concatenate([raizes2(s[:3]) for s in sos])
fig, ax = plt.subplots(2, len(bits), figsize=(4.2 * len(bits), 8.5))
for i, forma in enumerate(["cascata", "direta"]):
    for j, (nb, pq, zq, Hq, estado, saida) in enumerate(res_q[forma]):
        a_ = ax[i, j]
        a_.plot(np.cos(th), np.sin(th), "k--", lw=0.8)
        a_.plot(p0.real, p0.imag, "x", color="gray", ms=8, mew=2, alpha=0.6, label="polos orig.")
        a_.plot(z0.real, z0.imag, "o", mfc="none", mec="gray", ms=8, alpha=0.6, label="zeros orig.")
        a_.plot(pq.real, pq.imag, "rx", ms=8, mew=2, label="polos quant.")
        if len(zq):
            a_.plot(zq.real, zq.imag, "bo", mfc="none", ms=6, label="zeros quant.")
        a_.set_aspect("equal")
        a_.set_xlim(-1.6, 1.6)
        a_.set_ylim(-1.6, 1.6)
        a_.grid(alpha=0.3)
        a_.set_title(f"{forma}, b = {nb}" + {"sim": "", "marg.": " - MARG.", "NÃO": " - INSTÁVEL"}[estado],
                     color="k" if estado == "sim" else "r")
        if i == 0 and j == 0:
            a_.legend(fontsize=7, loc="lower left")
fig.tight_layout()
plt.savefig("Q5_09.png", dpi=110)
plt.show()

for forma in ["cascata", "direta"]:
    for nb, pq, zq, Hq, estado, saida in res_q[forma]:
        if estado == "sim" and saida is not None and np.all(np.isfinite(saida)) and np.max(np.abs(saida)) < 10:
            print(f"Áudio recuperado com coeficientes quantizados ({forma}, b = {nb}, sigma^2 = {s2_ref}):")
            display(Audio(saida, rate=fs, autoplay=False))
```

(a) A SNR de entrada é de 4,35 dB para sigma^2 = 0,01, -4,33 dB para 0,1 e -14,18 dB para 1. Com 0,01 a forma de onda original ainda é reconhecível. Com 0,1 e 1 o ruído domina, pois seu desvio padrão supera o valor RMS do sinal (0,196). No espectro, as interferências aparecem como raias em 100 Hz e 2000 Hz sobre um patamar plano de ruído, que sobe 10 dB a cada década de variância. Ao ouvir, nota-se um zumbido grave, um apito agudo e um chiado. Com sigma^2 = 1 a música fica quase mascarada.

(b) 95% da energia de x(t) está abaixo de 1719 Hz. As interferências são tons de frequência fixa e podem ser removidas com filtros notch. Do ruído branco, só a parte acima da banda do sinal pode ser removida, com um passa-baixas. Passa-altas e passa-faixas não são necessários. A limitação é que o ruído dentro da banda da música não pode ser separado por filtragem: com corte em 1800 Hz, só 56% da potência do ruído é removível.

(c) O sistema é uma cascata de 5 blocos básicos de 2ª ordem, montados a partir dos polos e zeros: um notch em 100 Hz, um notch em 2000 Hz e um passa-baixas de 6ª ordem com corte em 1800 Hz, formado por 3 blocos passa-baixas G (z + 1)^2 / (z^2 - 2r cos(theta) z + r^2). Para que a banda passante fique plana, os polos desses três blocos foram posicionados como os de um Butterworth: os polos analógicos foram calculados e levados ao plano z pela transformação bilinear, o que dá r = 0,163, 0,425 e 0,771 (ângulos de 1214, 1701 e 1792 Hz). Os notches têm zeros sobre o círculo unitário e polos com r = 0,99. A frequência de corte cobre 95% da energia da música. Cortes menores aumentariam a SNR, mas deixariam o som abafado. Para r, o valor 0,99 dá um entalhe de cerca de 26 Hz, com pouca distorção do sinal e transitório curto. Com r = 0,9 o entalhe distorceria a música e, com r = 0,999, o transitório seria mais longo e o bloco mais sensível à quantização.

(d) No sinal recuperado, as raias de 100 Hz e 2000 Hz desaparecem e o ruído acima de 1800 Hz é fortemente atenuado. Abaixo de 1800 Hz o ruído permanece no mesmo nível. No tempo, com sigma^2 = 0,01 o sinal recuperado acompanha x(t). Com 0,1 e 1 a forma de onda continua dominada pelo ruído.

(e) A SNR de saída é de 6,05 dB, -0,97 dB e -10,51 dB, com ganhos de 1,7, 3,4 e 3,7 dB. O MSE cai de 0,014 para 0,010, de 0,104 para 0,048 e de 1,009 para 0,433. O ruído residual fica em cerca de 43% do ruído de entrada nos três casos, e as interferências são praticamente eliminadas. O ganho é limitado pelo ruído dentro da banda e, para ruído pequeno, também pela distorção introduzida pelo filtro (perda acima de 1,8 kHz e fase não linear).

(f) Nos áudios recuperados, o zumbido e o apito somem completamente. O chiado fica mais grave, mas continua presente, e a música fica levemente abafada. Com sigma^2 = 0,01 o áudio é bem inteligível. Com 0,1 a música é reconhecível sob um chiado forte. Com 1 ela continua quase mascarada.

(g) A forma direta, de ordem 10, é instável com 2, 4 e 8 bits e só funciona com 16 bits. A cascata é marginalmente estável com 2 e 4 bits. Com 8 bits ela é estável, mas o notch de 100 Hz se desfaz: como seus polos e zeros estão muito próximos de z = 1, o arredondamento desloca os zeros para DC e cria uma ressonância de +18 dB perto de 165 Hz. Com isso, a SNR de saída cai para 1,9 dB. Com 16 bits as duas formas reproduzem o filtro original. A cascata é mais robusta porque cada seção tem um único par de polos, com baixa sensibilidade e estabilidade fácil de verificar. Os blocos de baixa frequência são os que exigem mais resolução.
