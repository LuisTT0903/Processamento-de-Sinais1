```python
# ============================================================
# Projeto de filtros FIR pelo método das janelas
# Implementação própria (sem scipy.signal) — fs = 20 kHz, M = 20, 50, 100
# Janelas: retangular, triangular, Bartlett, Hann, Hamming, Blackman
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
try:
    from IPython.display import display
except ImportError:
    display = print

plt.rcParams['figure.dpi'] = 90

fs = 20000.0
ORDENS = [20, 50, 100]

def janela_retangular(M):
    return np.ones(M + 1)

def janela_triangular(M):
    # Triangular "sem zeros nas pontas": 1 - |n - M/2| / (M/2 + 1)
    n = np.arange(M + 1)
    return 1.0 - np.abs(n - M / 2) / (M / 2 + 1)

def janela_bartlett(M):
    # Bartlett: triangular com zeros nas pontas: 1 - |n - M/2| / (M/2)
    n = np.arange(M + 1)
    return 1.0 - np.abs(n - M / 2) / (M / 2)

def janela_hann(M):
    n = np.arange(M + 1)
    return 0.5 - 0.5 * np.cos(2 * np.pi * n / M)

def janela_hamming(M):
    n = np.arange(M + 1)
    return 0.54 - 0.46 * np.cos(2 * np.pi * n / M)

def janela_blackman(M):
    n = np.arange(M + 1)
    return (0.42 - 0.5 * np.cos(2 * np.pi * n / M)
            + 0.08 * np.cos(4 * np.pi * n / M))

JANELAS = {
    'Retangular': janela_retangular,
    'Triangular': janela_triangular,
    'Bartlett':   janela_bartlett,
    'Hann':       janela_hann,
    'Hamming':    janela_hamming,
    'Blackman':   janela_blackman,
}

def hd_passa_baixas(M, fc):
    """h_d[n] = sen(wc (n - M/2)) / (pi (n - M/2)), com wc = 2 pi fc / fs."""
    wc = 2 * np.pi * fc / fs
    k = np.arange(M + 1) - M / 2
    h = np.empty(M + 1)
    centro = (k == 0)
    h[~centro] = np.sin(wc * k[~centro]) / (np.pi * k[~centro])
    h[centro] = wc / np.pi            # limite de sen(wc k)/(pi k) quando k -> 0
    return h

def impulso_atrasado(M):
    d = np.zeros(M + 1)
    d[M // 2] = 1.0                   # delta[n - M/2]
    return d

def hd_passa_altas(M, fc):
    return impulso_atrasado(M) - hd_passa_baixas(M, fc)

def hd_passa_faixas(M, fc1, fc2):
    return hd_passa_baixas(M, fc2) - hd_passa_baixas(M, fc1)

def hd_rejeita_faixas(M, fc1, fc2):
    return impulso_atrasado(M) - hd_passa_faixas(M, fc1, fc2)

def projeta_fir(M, tipo, nome_janela, f1, f2=None):
    if tipo == 'pb':
        hd = hd_passa_baixas(M, f1)
    elif tipo == 'pa':
        hd = hd_passa_altas(M, f1)
    elif tipo == 'pf':
        hd = hd_passa_faixas(M, f1, f2)
    elif tipo == 'rf':
        hd = hd_rejeita_faixas(M, f1, f2)
    else:
        raise ValueError(tipo)
    w = JANELAS[nome_janela](M)
    return hd * w                     # h[n] = h_d[n] . w[n]

def resposta_frequencia(h, n_pontos=4096):
    """H(e^{jw}) = sum_n h[n] e^{-jwn}, avaliada diretamente (DTFT) em [0, fs/2]."""
    f = np.linspace(0, fs / 2, n_pontos)
    w = 2 * np.pi * f / fs
    n = np.arange(len(h))
    H = np.exp(-1j * np.outer(w, n)) @ h
    return f, H

def polos_zeros(h):
    """H(z) = sum h[n] z^{-n} = (h0 z^M + h1 z^{M-1} + ... + hM) / z^M.
    Zeros: raízes do polinômio do numerador; polos: M polos em z = 0."""
    M = len(h) - 1
    coef = np.array(h, dtype=float)
    tol = 1e-12 * np.max(np.abs(coef))
    coef[np.abs(coef) < tol] = 0.0
    # pontas nulas (janelas com w[0] = w[M] = 0): a ordem efetiva cai
    nz_ini = np.argmax(coef != 0)
    nz_fim = len(coef) - np.argmax(coef[::-1] != 0)
    nucleo = coef[nz_ini:nz_fim]
    zeros = np.roots(nucleo) if len(nucleo) > 1 else np.array([])
    # zeros extras em z = 0 vindos das amostras nulas no fim
    zeros = np.concatenate([zeros, np.zeros(len(coef) - nz_fim)])
    polos = np.zeros(M - nz_ini)  # cancelamento das amostras nulas no início
    return zeros, polos

def plota_polos_zeros(ax, h, titulo=''):
    zeros, polos = polos_zeros(h)
    t = np.linspace(0, 2 * np.pi, 400)
    ax.plot(np.cos(t), np.sin(t), 'k--', lw=0.8)
    ax.plot(zeros.real, zeros.imag, 'o', mfc='none', mec='C0', ms=5, label='zeros')
    if len(polos):
        ax.plot(0, 0, 'x', color='C3', ms=9, mew=2, label='polos')
        ax.annotate(f'{len(polos)}', (0.06, 0.06), color='C3', fontsize=9)
    lim = min(max(1.3, 1.1 * np.max(np.abs(zeros)) if len(zeros) else 1.3), 3.0)
    fora = np.sum(np.abs(zeros) > lim)
    ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim)
    ax.set_aspect('equal'); ax.grid(alpha=0.3)
    ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
    ax.set_title(titulo + (f'  ({fora} zero(s) fora do quadro)' if fora else ''), fontsize=9)

def painel_filtro(tipo, M, f1, f2, nome_filtro):
    """Uma figura por (filtro, M): linhas = janelas; colunas = h[n], |H| em dB, polos/zeros."""
    fig, axs = plt.subplots(len(JANELAS), 3, figsize=(15, 3.0 * len(JANELAS)),
                            gridspec_kw={'width_ratios': [1.2, 1.6, 0.9]})
    for i, nome in enumerate(JANELAS):
        h = projeta_fir(M, tipo, nome, f1, f2)
        f, H = resposta_frequencia(h)
        HdB = 20 * np.log10(np.abs(H) + 1e-12)
        # resposta ao impulso
        ax = axs[i, 0]
        ax.stem(np.arange(M + 1), h, basefmt='k-')
        ax.set_title(f'{nome}: h[n]', fontsize=10); ax.set_xlabel('n'); ax.grid(alpha=0.3)
        # módulo em dB
        ax = axs[i, 1]
        ax.plot(f, HdB, lw=1.2)
        for fc in [f1] + ([f2] if f2 else []):
            ax.axvline(fc, color='r', ls='--', lw=0.8)
        ax.axhline(-6, color='gray', ls=':', lw=0.8)
        ax.set_ylim(-120, 10); ax.set_xlim(0, fs / 2)
        ax.set_title(f'{nome}: |H(f)| (dB)', fontsize=10); ax.set_xlabel('f (Hz)'); ax.grid(alpha=0.3)
        # polos e zeros
        plota_polos_zeros(axs[i, 2], h, f'{nome}: polos e zeros')
    fig.suptitle(f'{nome_filtro} — M = {M}', fontsize=14, y=1.0)
    fig.tight_layout(); plt.show()

def compara_janelas(tipo, f1, f2, nome_filtro, zoom):
    """Sobrepõe as 6 janelas para cada M (escala dB e zoom linear)."""
    fig, axs = plt.subplots(2, len(ORDENS), figsize=(16, 8))
    for j, M in enumerate(ORDENS):
        for nome in JANELAS:
            f, H = resposta_frequencia(projeta_fir(M, tipo, nome, f1, f2))
            axs[0, j].plot(f, 20 * np.log10(np.abs(H) + 1e-12), lw=1, label=nome)
            axs[1, j].plot(f, np.abs(H), lw=1, label=nome)
        for ax in axs[:, j]:
            for fc in [f1] + ([f2] if f2 else []):
                ax.axvline(fc, color='r', ls='--', lw=0.8)
            ax.grid(alpha=0.3); ax.set_xlabel('f (Hz)')
        axs[0, j].set_ylim(-120, 10); axs[0, j].set_xlim(0, fs / 2)
        axs[0, j].set_title(f'M = {M} — |H| (dB)')
        axs[1, j].set_xlim(*zoom); axs[1, j].set_title(f'M = {M} — |H| linear (zoom)')
    axs[0, 0].legend(fontsize=8)
    fig.suptitle(f'{nome_filtro}: comparação entre janelas', fontsize=14)
    fig.tight_layout(); plt.show()

def tabela_ganhos(tipo, f1, f2, freqs):
    """Ganho |H| em dB nas frequências de teste, para todas as janelas e ordens."""
    linhas = {}
    for M in ORDENS:
        for nome in JANELAS:
            h = projeta_fir(M, tipo, nome, f1, f2)
            n = np.arange(M + 1)
            H = [np.sum(h * np.exp(-2j * np.pi * fr / fs * n)) for fr in freqs]
            linhas[(M, nome)] = [20 * np.log10(abs(v) + 1e-15) for v in H]
    df = pd.DataFrame(linhas, index=[f'{fr} Hz' for fr in freqs]).T
    df.index.names = ['M', 'Janela']
    return df.round(1)

def metricas_janela(w):
    f, W = resposta_frequencia(w, 16384)
    m = np.abs(W) / np.abs(W[0])
    i = 1
    while i < len(m) - 1 and not (m[i] <= m[i - 1] and m[i] <= m[i + 1]):
        i += 1                                   # primeiro nulo
    return 2 * f[i], 20 * np.log10(np.max(m[i:]))  # largura do lóbulo principal, pico lateral

M = 50
fig, axs = plt.subplots(1, 2, figsize=(15, 4.5))
for nome, jan in JANELAS.items():
    w = jan(M)
    axs[0].plot(np.arange(M + 1), w, '.-', label=nome)
    f, W = resposta_frequencia(w, 8192)
    axs[1].plot(f, 20 * np.log10(np.abs(W) / np.abs(W[0]) + 1e-12), lw=1, label=nome)
axs[0].set_title(f'Janelas no tempo (M = {M})'); axs[0].set_xlabel('n'); axs[0].legend(); axs[0].grid(alpha=0.3)
axs[1].set_title(f'Espectro normalizado das janelas (M = {M})'); axs[1].set_xlabel('f (Hz)')
axs[1].set_ylim(-120, 5); axs[1].set_xlim(0, 5000); axs[1].grid(alpha=0.3); axs[1].legend()
plt.tight_layout(); plt.show()

tab = {}
for M in ORDENS:
    for nome, jan in JANELAS.items():
        lp, ls = metricas_janela(jan(M))
        tab[(M, nome)] = {'Lóbulo principal (Hz)': round(lp), 'Pico do lóbulo lateral (dB)': round(ls, 1)}
print('Métricas das janelas:')
display(pd.DataFrame(tab).T)

# ==================== (a) PASSA-BAIXAS ====================
painel_filtro('pb', 20, 1000, None, '(a) Passa-baixas, fc = 1000 Hz')

painel_filtro('pb', 50, 1000, None, '(a) Passa-baixas, fc = 1000 Hz')

painel_filtro('pb', 100, 1000, None, '(a) Passa-baixas, fc = 1000 Hz')

compara_janelas('pb', 1000, None, '(a) Passa-baixas, fc = 1000 Hz', zoom=(0, 4000))

print('Ganho |H(f)| em dB nas frequências de teste:')
display(tabela_ganhos('pb', 1000, None, [0, 1000, 1500, 2000, 3000, 5000]))

# ==================== (b) PASSA-ALTAS ====================
painel_filtro('pa', 20, 2000, None, '(b) Passa-altas, fc = 2000 Hz')

painel_filtro('pa', 50, 2000, None, '(b) Passa-altas, fc = 2000 Hz')

painel_filtro('pa', 100, 2000, None, '(b) Passa-altas, fc = 2000 Hz')

compara_janelas('pa', 2000, None, '(b) Passa-altas, fc = 2000 Hz', zoom=(0, 5000))

print('Ganho |H(f)| em dB nas frequências de teste:')
display(tabela_ganhos('pa', 2000, None, [0, 500, 1000, 2000, 3000, 8000]))

# ==================== (c) PASSA-FAIXAS ====================
painel_filtro('pf', 20, 500, 2000, '(c) Passa-faixas, fc1 = 500 Hz e fc2 = 2000 Hz')

painel_filtro('pf', 50, 500, 2000, '(c) Passa-faixas, fc1 = 500 Hz e fc2 = 2000 Hz')

painel_filtro('pf', 100, 500, 2000, '(c) Passa-faixas, fc1 = 500 Hz e fc2 = 2000 Hz')

compara_janelas('pf', 500, 2000, '(c) Passa-faixas, fc1 = 500 Hz e fc2 = 2000 Hz', zoom=(0, 5000))

print('Ganho |H(f)| em dB nas frequências de teste:')
display(tabela_ganhos('pf', 500, 2000, [0, 250, 500, 1250, 2000, 3000, 5000]))

# ==================== (d) REJEITA-FAIXAS ====================
painel_filtro('rf', 20, 1000, 2500, '(d) Rejeita-faixas, fc1 = 1000 Hz e fc2 = 2500 Hz')

painel_filtro('rf', 50, 1000, 2500, '(d) Rejeita-faixas, fc1 = 1000 Hz e fc2 = 2500 Hz')

painel_filtro('rf', 100, 1000, 2500, '(d) Rejeita-faixas, fc1 = 1000 Hz e fc2 = 2500 Hz')

compara_janelas('rf', 1000, 2500, '(d) Rejeita-faixas, fc1 = 1000 Hz e fc2 = 2500 Hz', zoom=(0, 5000))

print('Ganho |H(f)| em dB nas frequências de teste:')
display(tabela_ganhos('rf', 1000, 2500, [0, 500, 1000, 1750, 2500, 3500, 8000]))

# ==================== FASE LINEAR ====================
fig, axs = plt.subplots(1, 2, figsize=(15, 4))
for M in ORDENS:
    h = projeta_fir(M, 'pb', 'Hamming', 1000)
    f, H = resposta_frequencia(h)
    fase = np.unwrap(np.angle(H))
    axs[0].plot(f, fase, label=f'M = {M}')
    # atraso de grupo numérico: -d(fase)/d(omega), só na banda de passagem
    w = 2 * np.pi * f / fs
    tau = -np.diff(fase) / np.diff(w)
    passa = f[:-1] < 800
    axs[1].plot(f[:-1][passa], tau[passa], label=f'M = {M}  (M/2 = {M // 2})')
axs[0].set_title('Fase desenrolada (passa-baixas, Hamming)'); axs[0].set_xlabel('f (Hz)'); axs[0].set_ylabel('rad')
axs[1].set_title('Atraso de grupo na banda de passagem'); axs[1].set_xlabel('f (Hz)'); axs[1].set_ylabel('amostras')
for ax in axs: ax.grid(alpha=0.3); ax.legend()
plt.tight_layout(); plt.show()
```

Métricas das janelas

Os resultados seguem o que a teoria prevê. A retangular tem o lóbulo principal mais estreito (≈ 784 Hz para M = 50), mas o pior pico de lóbulo lateral (≈ −13 dB). Na Triangular e na Bartlett o lóbulo lateral cai para ≈ −26 dB, na Hann para ≈ −31,5 dB, na Hamming para ≈ −42 dB e na Blackman para ≈ −58 dB. Em troca, o lóbulo principal fica cerca de 2× mais largo (3× na Blackman). A largura do lóbulo principal é inversamente proporcional a M: dobrar M reduz a largura à metade. Já o pico lateral não muda com M, porque depende só do formato da janela.

(a) Passa-baixas, fc = 1000 Hz

Em todos os casos o ganho em fc fica perto de −6 dB (|H| ≈ 0,5), como esperado no método das janelas. Com M = 20 a transição é muito larga. A Blackman, por exemplo, já atenua ≈ 2,9 dB em DC e a retangular apresenta um ripple de +1,4 dB (fenômeno de Gibbs). Aumentando M, a transição fica mais estreita e a rejeição melhora. Com M = 100, a retangular e a Bartlett ficam limitadas a ≈ −45 dB na banda de rejeição, enquanto Hann, Hamming e Blackman passam de −68 dB (a Blackman chega a ≈ −96 dB acima de 3 kHz). A retangular é a que corta mais rápido logo após fc (−46,8 dB em 1500 Hz com M = 100), mas os lóbulos laterais limitam a atenuação mais adiante.

(b) Passa-altas, fc = 2000 Hz

O comportamento espelha o do passa-baixas. Com M = 20, nenhuma janela rejeita bem as baixas frequências (de −20 a −48 dB em DC). Com M = 100, a rejeição em 0–1000 Hz passa de −55 dB nas janelas Hann, Hamming e Blackman, e fica em torno de −32 dB na retangular e na triangular. A banda de passagem (3–8 kHz) fica praticamente plana (≈ 0 dB) a partir de M = 50.

(c) Passa-faixas, fc1 = 500 Hz e fc2 = 2000 Hz

A banda é estreita em relação à resolução de um filtro com M = 20, então nenhuma janela chega a 0 dB no centro da faixa (1250 Hz). As janelas mais suaves (Blackman, com −4,4 dB) praticamente não separam a banda de passagem das bandas de rejeição. Isso mostra que a transição depende de M e da janela, e que uma ordem baixa não basta para bordas tão próximas. Com M = 50 e 100, o ganho no centro fica em ≈ 0 dB, e as bordas ficam em −6 dB. A rejeição acima de 3 kHz passa de −60 dB nas janelas Hann, Hamming e Blackman. Abaixo de 500 Hz a rejeição é menor (de −20 a −50 dB), porque a faixa de transição inferior é estreita.

(d) Rejeita-faixas, fc1 = 1000 Hz e fc2 = 2500 Hz

Com M = 20 o “notch” é raso (de −7 a −20 dB em 1750 Hz), e as janelas suaves ainda atenuam a banda de passagem perto de DC (de −1 a −3 dB). Com o aumento de M, a rejeição no centro aumenta bastante. Com M = 100, a Blackman chega a ≈ −72 dB e a Hann a ≈ −60 dB, contra ≈ −30 dB da retangular. As bandas de passagem (0–500 Hz e acima de 3,5 kHz) ficam em ≈ 0 dB.

Polos e zeros

Todos os filtros têm apenas polos em z = 0 (M polos, ou menos nas janelas que zeram as pontas, como Bartlett e Hann). Por isso são sempre estáveis. Os zeros aparecem em pares recíprocos (z₀ e 1/z₀*), o que é consequência da simetria de h[n] e garante fase linear. Os zeros sobre a circunferência unitária ficam nas bandas de rejeição. Os zeros fora da circunferência acompanham os recíprocos de dentro.

Fase linear

A fase desenrolada é uma reta na banda de passagem, e o atraso de grupo é constante e igual a M/2 amostras (10, 25 e 50). Isso confirma que os filtros FIR simétricos têm fase linear. O custo é o atraso: com M = 100 são 50 amostras, ou 2,5 ms a 20 kHz.

Conclusão

A escolha da janela é um compromisso entre largura da transição e atenuação na banda de rejeição. A retangular dá a transição mais estreita e a pior rejeição. A Blackman dá a maior rejeição e a transição mais larga. A Hamming e a Hann são o meio-termo. Aumentar M estreita a transição em qualquer janela, mas não melhora o pico do lóbulo lateral, que depende só do tipo de janela. Também aumenta o atraso, que é sempre M/2 amostras.
