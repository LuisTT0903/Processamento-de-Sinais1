```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

fs = 20000
r_vals = [0.5, 0.7, 0.8, 0.9, 0.95, 0.99]
N_freq = 8192


def bloco(tipo, fc, r):
    th = 2 * np.pi * fc / fs
    a = np.array([1.0, -2 * r * np.cos(th), r ** 2])
    if tipo == "pb":
        b = np.array([1.0, 2.0, 1.0])
        g = np.polyval(a, 1.0) / np.polyval(b, 1.0)
    elif tipo == "pa":
        b = np.array([1.0, -2.0, 1.0])
        g = np.polyval(a, -1.0) / np.polyval(b, -1.0)
    elif tipo == "pf":
        b = np.array([1.0, 0.0, -1.0])
        z0 = np.exp(1j * th)
        g = abs(np.polyval(a, z0) / np.polyval(b, z0))
    elif tipo == "notch":
        b = np.array([1.0, -2 * np.cos(th), 1.0])
        g = np.polyval(a, 1.0) / np.polyval(b, 1.0)
    return g * b, a


def tf_texto(b, a):
    def poli(c):
        return f"{c[0]:.4f} z^2 {c[1]:+.4f} z {c[2]:+.4f}"
    return f"H(z) = ({poli(b)}) / ({poli(a)})"


def cruzamentos(f, mag, nivel):
    acima = mag >= nivel
    idx = np.where(np.diff(acima.astype(int)) != 0)[0]
    pts = []
    for i in idx:
        f1, f2, m1, m2 = f[i], f[i + 1], mag[i], mag[i + 1]
        pts.append(f1 + (nivel - m1) * (f2 - f1) / (m2 - m1))
    return pts


def metricas(tipo, fc, f, H):
    mag = np.abs(H)
    mdb = 20 * np.log10(mag + 1e-15)
    k = np.argmax(mag)
    res = {"f_pico": f[k], "G_pico_dB": mdb[k]}
    if tipo == "pb":
        c = cruzamentos(f, mag, 1 / np.sqrt(2))
        res["f_-3dB"] = c[-1] if c else np.nan
        res["BW"] = res["f_-3dB"]
    elif tipo == "pa":
        c = cruzamentos(f, mag, 1 / np.sqrt(2))
        res["f_-3dB"] = c[0] if c else np.nan
        res["BW"] = fs / 2 - res["f_-3dB"]
    elif tipo == "pf":
        c = cruzamentos(f, mag, mag[k] / np.sqrt(2))
        f1 = max([x for x in c if x < f[k]], default=0.0)
        f2 = min([x for x in c if x > f[k]], default=fs / 2)
        res["f1"], res["f2"] = f1, f2
        res["BW"] = f2 - f1
        res["Q"] = f[k] / res["BW"]
    elif tipo == "notch":
        c = cruzamentos(f, mag, 1 / np.sqrt(2))
        f1 = max([x for x in c if x < fc], default=np.nan)
        f2 = min([x for x in c if x > fc], default=np.nan)
        res["f1"], res["f2"] = f1, f2
        res["BW"] = f2 - f1
        res["G_nyq_dB"] = mdb[-1]
        res["G_fc_dB"] = np.nan
    return res


def analisar(tipo, fc, titulo):
    print("=" * 90)
    print(f"{titulo}  |  fc = {fc} Hz  |  theta = 2*pi*fc/fs = {2*fc/fs:.2f}*pi rad")
    print("=" * 90)
    cores = plt.cm.viridis(np.linspace(0, 0.9, len(r_vals)))
    fig, ax = plt.subplots(2, 2, figsize=(14, 10))
    linhas = []
    for r, cor in zip(r_vals, cores):
        b, a = bloco(tipo, fc, r)
        f, H = signal.freqz(b, a, worN=N_freq, fs=fs)
        z, p, _ = signal.tf2zpk(b, a)
        m = metricas(tipo, fc, f, H)
        if tipo == "notch":
            z0 = np.exp(2j * np.pi * fc / fs)
            m["G_fc"] = abs(np.polyval(b, z0) / np.polyval(a, z0))
        m["r"] = r
        m["|p|"] = np.max(np.abs(p))
        m["ang_p_Hz"] = abs(np.angle(p[0])) * fs / (2 * np.pi)
        linhas.append(m)
        print(f"r = {r}:  {tf_texto(b, a)}")
        print(f"         polos = {np.round(p, 4)}   zeros = {np.round(z, 4)}")
        ax[0, 0].plot(f, 20 * np.log10(np.abs(H) + 1e-15), color=cor, label=f"r = {r}")
        ax[0, 1].plot(f, np.unwrap(np.angle(H)), color=cor, label=f"r = {r}")
        ax[1, 0].plot(np.real(p), np.imag(p), "x", color=cor, ms=10, mew=2, label=f"polos r = {r}")
        _, h = signal.dimpulse((b, a, 1 / fs), n=150)
        ax[1, 1].plot(np.arange(150), np.squeeze(h), color=cor, label=f"r = {r}")
    ax[1, 0].plot(np.real(z), np.imag(z), "o", mfc="none", mec="red", ms=12, mew=2, label="zeros")
    t = np.linspace(0, 2 * np.pi, 400)
    ax[1, 0].plot(np.cos(t), np.sin(t), "k--", lw=0.8)
    ax[1, 0].axhline(0, color="gray", lw=0.5)
    ax[1, 0].axvline(0, color="gray", lw=0.5)
    ax[1, 0].set_aspect("equal")
    ax[1, 0].set_xlim(-1.3, 1.3)
    ax[1, 0].set_ylim(-1.3, 1.3)
    ax[0, 0].axvline(fc, color="red", ls=":", label=f"fc = {fc} Hz")
    ax[0, 0].axhline(-3, color="gray", ls=":")
    ax[0, 0].set_ylim(-60, 25)
    ax[0, 1].axvline(fc, color="red", ls=":")
    ax[0, 0].set_title("Módulo da resposta em frequência")
    ax[0, 0].set_xlabel("f [Hz]")
    ax[0, 0].set_ylabel("|H| [dB]")
    ax[0, 1].set_title("Fase da resposta em frequência")
    ax[0, 1].set_xlabel("f [Hz]")
    ax[0, 1].set_ylabel("fase [rad]")
    ax[1, 0].set_title("Diagrama de polos e zeros")
    ax[1, 0].set_xlabel("Re{z}")
    ax[1, 0].set_ylabel("Im{z}")
    ax[1, 1].set_title("Resposta ao impulso")
    ax[1, 1].set_xlabel("n")
    ax[1, 1].set_ylabel("h[n]")
    for a_ in ax.flat:
        a_.grid(alpha=0.3)
    ax[0, 0].legend(fontsize=8)
    ax[0, 1].legend(fontsize=8)
    ax[1, 1].legend(fontsize=8)
    ax[1, 0].legend(fontsize=8, loc="upper left", bbox_to_anchor=(1.02, 1))
    fig.suptitle(f"{titulo} - fc = {fc} Hz, fs = {fs} Hz", fontsize=14)
    fig.tight_layout()
    plt.savefig(f"{tipo}.png", dpi=110)
    plt.show()
    print()
    if tipo == "pb":
        print(f"{'r':>5} {'|p|':>6} {'ang polo [Hz]':>14} {'f pico [Hz]':>12} {'pico [dB]':>10} {'f -3dB [Hz]':>12}")
        for m in linhas:
            print(f"{m['r']:>5} {m['|p|']:>6.3f} {m['ang_p_Hz']:>14.1f} {m['f_pico']:>12.1f} {m['G_pico_dB']:>10.2f} {m['f_-3dB']:>12.1f}")
    elif tipo == "pa":
        print(f"{'r':>5} {'|p|':>6} {'ang polo [Hz]':>14} {'f pico [Hz]':>12} {'pico [dB]':>10} {'f -3dB [Hz]':>12} {'BW [Hz]':>9}")
        for m in linhas:
            print(f"{m['r']:>5} {m['|p|']:>6.3f} {m['ang_p_Hz']:>14.1f} {m['f_pico']:>12.1f} {m['G_pico_dB']:>10.2f} {m['f_-3dB']:>12.1f} {m['BW']:>9.1f}")
    elif tipo == "pf":
        print(f"{'r':>5} {'|p|':>6} {'f pico [Hz]':>12} {'f1 [Hz]':>9} {'f2 [Hz]':>9} {'BW [Hz]':>9} {'BW teor [Hz]':>12} {'Q':>7}")
        for m in linhas:
            bw_t = (1 - m['r']) * fs / np.pi
            print(f"{m['r']:>5} {m['|p|']:>6.3f} {m['f_pico']:>12.1f} {m['f1']:>9.1f} {m['f2']:>9.1f} {m['BW']:>9.1f} {bw_t:>12.1f} {m['Q']:>7.2f}")
    elif tipo == "notch":
        print(f"{'r':>5} {'|p|':>6} {'|H(fc)|':>11} {'f1 [Hz]':>9} {'f2 [Hz]':>9} {'BW rej [Hz]':>12} {'BW teor [Hz]':>12} {'G(fs/2) [dB]':>13}")
        for m in linhas:
            bw_t = (1 - m['r']) * fs / np.pi
            print(f"{m['r']:>5} {m['|p|']:>6.3f} {m['G_fc']:>11.1e} {m['f1']:>9.1f} {m['f2']:>9.1f} {m['BW']:>12.1f} {bw_t:>12.1f} {m['G_nyq_dB']:>13.2f}")
    print()


def estabilidade(tipo, fc, titulo):
    fig, ax = plt.subplots(1, 3, figsize=(15, 4))
    for i, r in enumerate([0.99, 1.0, 1.01]):
        b, a = bloco(tipo, fc, r)
        h = signal.lfilter(b, a, np.r_[1.0, np.zeros(799)])
        ax[i].plot(h, lw=0.8)
        ax[i].set_title(f"{titulo}: h[n] com r = {r}")
        ax[i].set_xlabel("n")
        ax[i].grid(alpha=0.3)
    fig.tight_layout()
    plt.savefig("estabilidade.png", dpi=110)
    plt.show()


analisar("pb", 1000, "(a) Passa-baixas")
analisar("pa", 2000, "(b) Passa-altas")
analisar("pf", 7000, "(c) Passa-faixa")
analisar("notch", 3000, "(d) Notch")
estabilidade("pf", 7000, "Passa-faixa")
```

Os quatro filtros usam o mesmo denominador, z^2 - 2r cos(theta) z + r^2, com polos em r e^(±j theta) e theta = 2 pi fc / fs. Só os zeros mudam de um tipo para outro. O ganho G normaliza cada filtro para 0 dB na sua banda passante.

(a) Passa-baixas, H(z) = G (z + 1)^2 / (z^2 - 2r cos(theta) z + r^2). O zero duplo em z = -1 anula fs/2. Para r = 0,5 e 0,7 a resposta é monotônica, com corte em 1648 e 1370 Hz. A partir de r = 0,8 surge uma ressonância perto de 1000 Hz, que chega a 23,7 dB com r = 0,99. A transição fica mais íngreme, mas a banda passante é distorcida pelo pico.

(b) Passa-altas, H(z) = G (z - 1)^2 / (z^2 - 2r cos(theta) z + r^2). O zero duplo em z = 1 anula DC. O comportamento espelha o do passa-baixas: sem pico para r = 0,5 (corte em 3058 Hz) e com pico crescente em direção a 2000 Hz, chegando a 29,3 dB com r = 0,99.

(c) Passa-faixas, H(z) = G (z^2 - 1) / (z^2 - 2r cos(theta) z + r^2). Os zeros em z = ±1 anulam DC e fs/2. A largura de banda cai de 3440 Hz (r = 0,5) para 64 Hz (r = 0,99), seguindo bem a aproximação (1 - r) fs / pi para r >= 0,8. O fator Q vai de 1,9 a 109.

(d) Notch, H(z) = G (z^2 - 2 cos(theta) z + 1) / (z^2 - 2r cos(theta) z + r^2). Os zeros estão sobre o círculo unitário, então o ganho em 3000 Hz é nulo para qualquer r. O parâmetro r só define a largura da rejeição, de 2513 Hz (r = 0,5) a 64 Hz (r = 0,99). Com r pequeno, o ganho fora do entalhe sobe (2,85 dB em fs/2 para r = 0,5).

Em todos os casos, aproximar r de 1 aumenta a seletividade, mas também aumenta a ressonância e o transitório da resposta ao impulso. O filtro é estável apenas para r < 1: com r = 1 a resposta ao impulso oscila sem decair e com r = 1,01 ela diverge.
